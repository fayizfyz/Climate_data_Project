```python
import pandas as pd
combined_ddf = pd.concat([df1, df2,df3,df4,df5,df6,df7,df8], axis=0)

#
```

```python
combined_ddf
```

```python
combined_ddf = combined_ddf.drop('height',axis=1)
```

```python
combined_ddf['utci_celsius'] = combined_ddf['utci'] - 273.15
```

```python
def classify_utci(utci_celsius):
    if utci_celsius > 46:
        return 'Extreme Heat Stress'
    elif 38 <= utci_celsius <= 46:
        return 'Very Strong Heat Stress'
    elif 32 <= utci_celsius < 38:
        return 'Strong Heat Stress'
    elif 26 <= utci_celsius < 32:
        return 'Moderate Heat Stress'
    elif 9 <= utci_celsius < 26:
        return 'No thermal stress'
    elif 0 <= utci_celsius < 9:
        return 'Slight Cold Stress'
    elif -13 <= utci_celsius < 0:
        return 'Moderate Cold Stress'
    elif -27 <= utci_celsius < -13:
        return 'Strong Cold Stress'
    elif -40 <= utci_celsius < -27:
        return 'Very Strong Cold Stress'
    else:
        return 'Extreme Cold Stress'

combined_ddf['stress_category'] = combined_ddf['utci_celsius'].apply(classify_utci)
```

```python
combined_ddf.to_csv('/content/drive/MyDrive/Work @Weclimb/Processed data/daily average/file(1950-2023).csv', index=False)
```

```python
import seaborn as sns

pivot_df = df.pivot_table(index='lat', columns='lon', values='utci_celsius', aggfunc='mean')
sns.heatmap(pivot_df, cmap='viridis')
plt.title('Spatial Heatmap of UTCI Index')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming your DataFrame is named 'df'
plt.figure(figsize=(12, 6))
sns.lineplot(x='time', y='utci', data=df)
plt.title('UTCI over Time')
plt.xlabel('Time')
plt.ylabel('UTCI')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(25, 12))
plt.plot(df['time'], df['utci_celsius'], marker='o', linestyle='-', color='b')
plt.title('Time Series Plot of UTC Index')
plt.xlabel('Time')
plt.ylabel('UTCI Index (Celsius)')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Convert 'time' column to datetime
df['time'] = pd.to_datetime(df['time'])

# Plotting
plt.figure(figsize=(15, 8))
plt.plot(df['time'], df['utci_celsius'], marker='o', linestyle='-', color='g')

# Improve x-axis labels
plt.gca().xaxis.set_major_locator(mdates.YearLocator())
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

# Set x-axis limits to start from the earliest date in the data
plt.xlim(df['time'].min(), df['time'].max())  # Set the x-axis limits based on the minimum and maximum dates in the DataFrame

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)  # Adjust the rotation angle as needed for better readability

plt.title('Time Series Plot of UTCI Index')
plt.xlabel('Time')
plt.ylabel('UTCI Index (Celsius)')
plt.grid(True)  # Add grid lines for better visualization
plt.tight_layout()  # Adjust layout to prevent clipping of labels
plt.show()
```
